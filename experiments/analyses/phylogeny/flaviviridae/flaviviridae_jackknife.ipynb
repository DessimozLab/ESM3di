# *Flaviviridae* glycoprotein fold trees: jackknife of the monophyly scores

How robust are the differences between the monophyly scores of the ESM3Di and AlphaFold2
trees of [flaviviridae_trees.ipynb](flaviviridae_trees.ipynb) to the choice of sequences?

**Method: paired delete-20% jackknife.** In each replicate a random 20% of the sequences is removed,
and the trees of both methods are rebuilt from the remaining 80%: the same subset for both,
so each replicate gives paired differences. 100 replicates per protein (E, E1, E2). The tips with an
unclassified code (..UN) are left out of the replicates, as of the scores in the figures: each is a group of its
own, whose position would otherwise decide part of the score.

The trees are rebuilt from the all-vs-all Foldseek results of `flaviviridae_trees.ipynb` with the
same steps as `esm3di foldtree` (distance matrix, QuickTree, negative branch lengths set to 0, MAD
rooting); 3Di prediction and Foldseek are not rerun. The search is exhaustive with `-e inf`, and a
pair's identity does not depend on the other sequences, so this equals rerunning the workflow on
the reduced set. Step 1 checks that the rebuild reproduces the workflow trees exactly.

**What it measures:** how much the scores and their differences depend on which sequences are in
the tree (taxon sampling). It does not capture uncertainty of the 3Di predictions or the
alignments, and replicates overlap in most of their sequences, so the percentile range is a
descriptive spread, not a confidence interval in the strict sense.

Run `flaviviridae_trees.ipynb` first. Results are cached in `RESULTS_CSV`; to recompute them, run on
a compute node and set `WORKERS` to the available cores. `flaviviridae_ultrametricity.ipynb` draws these replicates
in the figure `figures/flaviviridae_jackknife`.

```python
import os
import tempfile
from pathlib import Path

import pandas as pd
from Bio import Phylo

import sys
sys.path.insert(0, "../shared")  # code shared by the phylogeny analyses
from foldtree_resample import build_tree, jackknife, load_allvall
from monophyly import glycoprotein_labels, relabel_tree, score_monophyly

INPUT_DIR = Path("inputs")
OUTPUT_DIR = Path("/scratch/fweikert/phylogeny/flaviviridae")    # esm3di foldtree runs of flaviviridae_trees.ipynb
RESULTS_CSV = Path("results/flaviviridae_jackknife.csv")

REPLICATES = 100
FRACTION = 0.8      # share of sequences kept per replicate
SEED = 1
WORKERS = min(16, os.cpu_count())

DATASETS = {"fullglyco_E": "E", "fullglyco_E1": "E1", "fullglyco_E2": "E2"}
MODELS = {"esm3di": "ESM3Di", "af2": "AlphaFold2"}

QUICKTREE = next(OUTPUT_DIR.glob(".snakemake_conda/*/bin/quicktree"))   # installed by esm3di foldtree
labels = glycoprotein_labels({
    "E": INPUT_DIR / "refolded_fullglyco_E_aa.fas",
    "E1": INPUT_DIR / "refolded_fullglyco_E1_aa.fas",
    "E2": INPUT_DIR / "refolded_fullglyco_E2_aa.fas",
})
allvall = {dataset: {model: load_allvall(OUTPUT_DIR / f"{model}_{dataset}" / "allvall_1.csv") for model in MODELS}
           for dataset in DATASETS}


def classified(table):
    """The all-vs-all table without the tips with an unclassified code."""
    unclassified = {name for name in set(table["query"]) if labels[name][4:6] == "UN"}
    return table[~table["query"].isin(unclassified) & ~table["target"].isin(unclassified)]
```

## 1. Check: the rebuild reproduces the workflow trees

With all sequences kept, every rebuilt tree must have the same rooted clades and the same score as
the tree written by `esm3di foldtree`.

```python
def clades(path):
    return {frozenset(leaf.name for leaf in clade.get_terminals())
            for clade in Phylo.read(path, "newick").get_nonterminals()}


rows = []
for dataset in DATASETS:
    for model in MODELS:
        workflow_tree = OUTPUT_DIR / f"{model}_{dataset}" / f"{model}_struct_tree.PP.nwk.rooted"
        with tempfile.TemporaryDirectory() as workdir:
            rebuilt = build_tree(allvall[dataset][model], workdir, QUICKTREE)
            score = score_monophyly(relabel_tree(rebuilt, Path(workdir) / "rebuilt.nwk", labels), verbose=False)
            reference = score_monophyly(relabel_tree(workflow_tree, Path(workdir) / "workflow.nwk", labels),
                                        verbose=False)
            rows.append((DATASETS[dataset], MODELS[model], clades(rebuilt) == clades(workflow_tree),
                         score[0], reference[0], score[1]))
check = pd.DataFrame(rows, columns=["dataset", "method", "same_rooted_clades", "monophyletic_rebuilt",
                                    "monophyletic_workflow", "internal_nodes"])
assert check["same_rooted_clades"].all() and (check["monophyletic_rebuilt"] == check["monophyletic_workflow"]).all()
# Scores of the rebuilt trees of all classified sequences, the reference of the replicates
rows = []
for dataset in DATASETS:
    for model in MODELS:
        with tempfile.TemporaryDirectory() as workdir:
            rebuilt = build_tree(classified(allvall[dataset][model]), workdir, QUICKTREE)
            score = score_monophyly(relabel_tree(rebuilt, Path(workdir) / "rebuilt.nwk", labels), verbose=False)
            rows.append((DATASETS[dataset], MODELS[model], score[2]))
full_scores = pd.DataFrame(rows, columns=["dataset", "method", "fraction"])
check
```

## 2. Jackknife replicates

```python
if RESULTS_CSV.exists():
    replicates = pd.read_csv(RESULTS_CSV)
    print(f"Loaded {len(replicates)} rows from {RESULTS_CSV}")
else:
    tables = []
    for dataset, name in DATASETS.items():
        print(f"{name}: {REPLICATES} replicates on {WORKERS} workers")
        table = jackknife({MODELS[m]: classified(allvall[dataset][m]) for m in MODELS}, labels, QUICKTREE,
                          replicates=REPLICATES, fraction=FRACTION, seed=SEED, workers=WORKERS)
        tables.append(table.assign(dataset=name))
    replicates = pd.concat(tables, ignore_index=True)
    RESULTS_CSV.parent.mkdir(exist_ok=True)
    replicates.to_csv(RESULTS_CSV, index=False)
replicates.head()
```

## 3. Summary

Per protein: the full-data scores, the mean jackknife scores, and the paired differences of ESM3Di
from AlphaFold2 in the fraction of monophyletic internal nodes (mean, 2.5th–97.5th
percentile, and the share of replicates in which ESM3Di scores higher).

```python
paired = replicates.pivot_table(index=["dataset", "replicate"], columns="method", values="fraction")
summary = []
for name in DATASETS.values():
    rows = paired.loc[name]
    full = full_scores[full_scores["dataset"] == name].set_index("method")["fraction"]
    for other in ("AlphaFold2",):
        d = rows["ESM3Di"] - rows[other]
        summary.append({
            "protein": name, "ESM3Di minus": other,
            "full ESM3Di": full["ESM3Di"], "full other": full[other],
            "jackknife ESM3Di": rows["ESM3Di"].mean(), "jackknife other": rows[other].mean(),
            "mean difference": d.mean(), "2.5%": d.quantile(0.025), "97.5%": d.quantile(0.975),
            "ESM3Di higher": (d > 0).mean(), "equal": (d == 0).mean(), "other higher": (d < 0).mean(),
        })
summary = pd.DataFrame(summary).set_index(["protein", "ESM3Di minus"])
summary.round(4)
```

## 4. Summary with ultrametricity (Supplementary Table 2)

The same replicates, for the monophyly score and for ultrametricity, against the trees of Figure 2 and
Supplementary Fig. 5 (`results/flaviviridae_monophyly.csv`). These are built the same way, without the unclassified
sequences, which is checked below. Written to `results/flaviviridae_jackknife_summary.csv`.


```python
figure_scores = pd.read_csv("results/flaviviridae_monophyly.csv").set_index(["protein", "tree"])
rows = []
for name in DATASETS.values():
    full = figure_scores.xs(name, level="protein")
    check = full_scores[full_scores["dataset"] == name].set_index("method")["fraction"]
    for method in MODELS.values():
        assert abs(check[method] - full.loc[f"{method} FoldTree", "fraction_without_unclassified"]) < 1e-9, (name, method)
    for measure, column in (("fraction", "fraction_without_unclassified"), ("root_to_tip_cv", "root_to_tip_cv_without_unclassified")):
        paired = replicates[replicates["dataset"] == name].pivot(index="replicate", columns="method", values=measure)
        d = paired["ESM3Di"] - paired["AlphaFold2"]
        rows.append({"protein": name, "measure": measure, "full ESM3Di": full.loc["ESM3Di FoldTree", column],
                     "full AlphaFold2": full.loc["AlphaFold2 FoldTree", column], "median ESM3Di": paired["ESM3Di"].median(),
                     "median AlphaFold2": paired["AlphaFold2"].median(), "mean difference": d.mean(), "2.5%": d.quantile(0.025),
                     "97.5%": d.quantile(0.975), "ESM3Di higher": (d > 0).mean(), "equal": (d == 0).mean(), "ESM3Di lower": (d < 0).mean()})
jackknife_summary = pd.DataFrame(rows)
jackknife_summary.to_csv("results/flaviviridae_jackknife_summary.csv", index=False)
jackknife_summary.round(4)
```

## 5. Statistical significance of the differences

For each protein and measure: is the ESM3Di - AlphaFold2 difference of the trees of all sequences (the black marks of
the figure) distinguishable from 0, given how much it depends on which sequences were sampled?

**Test: delete-d jackknife z-test** (Shao & Wu 1989, *Ann. Stat.* 17:1176; Efron & Tibshirani 1993, ch. 11). The standard
error of the difference is the delete-d jackknife estimate from the paired replicates,
`se = sqrt((n - d) / (d * R) * sum((D_r - mean D)^2))`, for `d` of `n` sequences deleted per replicate and `R` replicates
(the same formula as `flaviviridae_ultrametricity.ipynb`). `z = D_all / se`, with `D_all` the difference of the trees of
all sequences; two-sided p from the normal distribution, Holm-corrected over the 6 tests (3 proteins x 2 measures).

**Why not a paired t-test, Wilcoxon or sign test over the 100 replicates.** These treat the replicates as 100
independent observations. They are not: they are subsets of the same 236, 189 or 188 sequences, and any two share
about 64% of them. The replicates vary less than the statistic itself would over new samples of sequences, by a factor
of about `sqrt((n - d) / d) = 2`, which the jackknife factor corrects, and a test over them gets smaller p-values the
more replicates are drawn, without any new data. The column `naive paired t p` shows this: p down to 1e-28.

**Conditions** (second table):

1. *Paired design.* Both methods are scored on the same subset in each replicate, so the difference per replicate is
   the right unit. Met.
2. *Smoothness.* The monophyly score is a count over a tree topology (`distinct differences`: 12-20 values in 100
   replicates), and both measures depend on the MAD root. The delete-1 jackknife fails for such non-smooth statistics;
   the delete-d jackknife is consistent for them when `d` grows faster than `sqrt(n)` (Shao & Wu 1989). Here
   `d` = 38-47 against `sqrt(n)` = 14-15. Met.
3. *Normality of the z statistic.* The Shapiro-Wilk test rejects normality of the replicate differences for most rows
   (ties of the monophyly score, skew of the CV), so the p-values are approximate. With p > 0.5 throughout, this
   does not decide anything.
4. *Stability of the standard error.* For a smooth statistic, deleting 20% and 50% estimate the same variance. Here
   they differ by up to a factor 2.4 (E ultrametricity), and the jackknife bias is larger than the standard error in
   3 of the 6 rows (`bias / se`). Both say the statistics change non-smoothly with the sequences; for example the CV of the AlphaFold2 E tree, 0.021 with all sequences, ranges from 0.015 to 0.067 when one group is left out
   (`flaviviridae_ultrametricity.ipynb`, test 3). The standard errors are therefore themselves uncertain; the
   table gives z for both, and neither is significant.
5. *Independence of the sequences.* The jackknife treats sequences as exchangeable draws; sequences of the same viral
   group are related. This makes the standard error too small (the test anti-conservative). It cannot turn a
   non-significant result into a significant one, so it does not affect the conclusion here.
6. *Monte Carlo error.* 100 random subsets instead of all of them; the relative error of the standard error is about
   `1 / sqrt(2R)` = 7%.

Written to `results/flaviviridae_jackknife_significance.csv`.

```python
import numpy as np
from scipy import stats


def holm(p):
    # Holm-adjusted p-values
    p = np.asarray(p, dtype=float)
    order = np.argsort(p)
    adjusted = np.minimum(1, np.maximum.accumulate((len(p) - np.arange(len(p))) * p[order]))
    result = np.empty_like(adjusted)
    result[order] = adjusted
    return result


full_differences = jackknife_summary.set_index(["protein", "measure"])
# standard errors of the delete-50% replicates of flaviviridae_ultrametricity.ipynb, as a check of stability
se_half = pd.read_csv("results/flaviviridae_ultrametricity_summary.csv").query("kept == 0.5").set_index(["protein", "measure"])["se"]
rows = []
for dataset, name in DATASETS.items():
    n = len(set.intersection(*(set(classified(allvall[dataset][model])["query"]) for model in MODELS)))
    group = replicates[replicates["dataset"] == name]
    deleted = n - group["tips"].iloc[0]
    for measure in ("fraction", "root_to_tip_cv"):
        paired = group.pivot(index="replicate", columns="method", values=measure)
        d = paired["ESM3Di"] - paired["AlphaFold2"]
        se = np.sqrt((n - deleted) / (deleted * len(d)) * ((d - d.mean()) ** 2).sum())
        full = full_differences.loc[(name, measure), "full ESM3Di"] - full_differences.loc[(name, measure), "full AlphaFold2"]
        rows.append({"protein": name, "measure": measure, "n": n, "deleted": deleted, "replicates": len(d),
                     "difference": full, "se": se, "z": full / se, "p": 2 * stats.norm.sf(abs(full / se)),
                     "naive paired t p": stats.ttest_1samp(d, 0).pvalue,
                     "deleted > sqrt(n)": deleted > np.sqrt(n), "distinct differences": d.nunique(),
                     "Shapiro p": stats.shapiro(d).pvalue,
                     "bias / se": (n - deleted) / deleted * (d.mean() - full) / se,
                     "se, 50% deleted": se_half.loc[(name, measure)],
                     "z, 50% deleted": full / se_half.loc[(name, measure)]})
significance = pd.DataFrame(rows)
significance.insert(significance.columns.get_loc("p") + 1, "p (Holm)", holm(significance["p"]))
significance.insert(significance.columns.get_loc("p (Holm)") + 1, "significant (0.05)", significance["p (Holm)"] < 0.05)
significance.to_csv("results/flaviviridae_jackknife_significance.csv", index=False)

display(significance[["protein", "measure", "difference", "se", "z", "p", "p (Holm)", "significant (0.05)",
                      "naive paired t p"]].round(4).assign(**{"naive paired t p": significance["naive paired t p"].map("{:.1e}".format)}))
display(significance[["protein", "measure", "n", "deleted", "deleted > sqrt(n)", "distinct differences", "Shapiro p",
                      "bias / se", "se", "se, 50% deleted", "z, 50% deleted"]].round(4))
```

**Result.** No difference between ESM3Di and AlphaFold2 is statistically significant, for either measure or any
protein: |z| < 0.7 and p > 0.5 before correction, p (Holm) = 1. With the delete-50% standard errors the conclusion is
the same. The share of replicates favouring one method (97% for E ultrametricity) is not a p-value: the replicates
overlap, and the trees of all sequences lie well within the spread that sequence sampling produces. So the two
methods give trees of indistinguishable monophyly and ultrametricity on these data, which is the claim the figure
supports.

## 6. Figure: the paired differences

The replicate differences themselves, ESM3Di minus AlphaFold2, against the line at no difference. The paired design
is the point: both trees of a replicate come from the same sequences, so the replicate-to-replicate variation cancels
and what is left is the difference between the methods. A distribution that covers the line means the two are not
separated by that measure on that dataset. The test of section 5 is not drawn: none of the differences is significant, which the caption states. Written to `figures/flaviviridae_jackknife_difference`.

```python
import sys

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

sys.path.insert(0, "../../../../benchmarks")   # the shared figure style
from pub_style import MM, MUTED_TEXT, WIDTH_SINGLE, save_pub, use_pub_style

use_pub_style()

BOX_COLOUR = "#3C5A75"
WHISKERS = (2.5, 97.5)      # the whiskers span 95% of the replicates
FULL_HALF = 0.37            # half-height of the mark for the tree of all sequences; the box is 0.25
ARROW_ROW = -0.93           # height of the "better" arrows, in rows, above the first box
ARROW_FRACTION = 0.26       # length of each arrow, as a fraction of the width of the axes
# measure, factor onto the units of the axis, axis label, the method favoured left and right of
# zero, and the limits of the axis (None: automatic; ultrametricity reaches further right than its
# data, so that the right-hand arrow fits). The direction differs between the two measures, which is
# the one thing a difference axis cannot show by itself, so it is on the arrows.
PANELS = [("Monophyly", "fraction", 100, "ESM3Di - AlphaFold2 (percentage points)",
           ("AlphaFold2", "ESM3Di"), None),
          ("Ultrametricity", "root_to_tip_cv", 1, "ESM3Di - AlphaFold2",
           ("ESM3Di", "AlphaFold2"), (-0.066, 0.026))]

summary = jackknife_summary.set_index(["protein", "measure"])
fig = plt.figure(figsize=(WIDTH_SINGLE, 70 * MM), layout="constrained")
axes = fig.subplots(len(PANELS), 1)

for ax, (title, measure, factor, xlabel, _, xlim) in zip(axes, PANELS):
    differences, full = [], []
    for name in DATASETS.values():
        paired = replicates[replicates["dataset"] == name].pivot(index="replicate", columns="method", values=measure)
        differences.append(factor * (paired["ESM3Di"] - paired["AlphaFold2"]).to_numpy())
        scores = summary.loc[(name, measure)]
        full.append(factor * (scores["full ESM3Di"] - scores["full AlphaFold2"]))

    # no difference: over the rows and the arrows, and above the boxes, so that the boxes do not hide
    # parts of it and it is drawn the same in both panels
    ax.plot([0, 0], [ARROW_ROW - 0.22, len(differences) - 0.55], color=MUTED_TEXT, lw=0.6, ls=(0, (2.5, 2)), zorder=6)
    ax.boxplot(differences, positions=range(len(differences)), orientation="horizontal", widths=0.5, whis=WHISKERS,
               showfliers=False, patch_artist=True, zorder=3,
               boxprops=dict(facecolor="#C9D6E1", edgecolor=BOX_COLOUR, linewidth=0.6),
               medianprops=dict(color=BOX_COLOUR, linewidth=1.2),
               whiskerprops=dict(color=BOX_COLOUR, linewidth=0.6),
               capprops=dict(color=BOX_COLOUR, linewidth=0.6))

    # where the difference of the trees of all sequences falls among the replicates; taller than the
    # box, so it is not taken for the median line inside it
    for row, difference in enumerate(full):
        ax.plot([difference, difference], [row - FULL_HALF, row + FULL_HALF], color="black", lw=0.9,
                solid_capstyle="butt", zorder=5)
    ax.set_ylim(len(differences) - 0.55, ARROW_ROW - 0.3)
    if xlim:
        ax.set_xlim(xlim)
    ax.set_yticks(range(len(differences)), DATASETS.values(), fontsize=6)
    ax.tick_params(axis="y", length=0, pad=2)
    ax.spines["left"].set_visible(False)
    ax.set_xlabel(xlabel)
    ax.set_title(title, loc="left", pad=3)

# A key for the two marks, on the title line of the first panel, where it covers nothing
axes[0].legend(handles=[Line2D([], [], color="black", lw=0.9, label="Tree of all sequences"),
                        Line2D([], [], color=MUTED_TEXT, lw=0.6, ls=(0, (2.5, 2)), label="No difference")],
               loc="lower right", bbox_to_anchor=(1, 1.0), ncol=2, frameon=False, fontsize=5.5,
               handlelength=1.4, handletextpad=0.4, columnspacing=1.2, borderpad=0, borderaxespad=0.2)

fig.canvas.draw()
fig.set_layout_engine("none")

# Which side favours which method: an arrow away from zero on each side, with its label on it
for ax, (*_, favoured, _) in zip(axes, PANELS):
    zero = ax.transAxes.inverted().transform(ax.transData.transform((0, 0)))[0]
    y = ax.transAxes.inverted().transform(ax.transData.transform((0, ARROW_ROW)))[1]
    for side, method in zip((-1, 1), favoured):
        start, end = zero + side * 0.012, zero + side * (0.012 + ARROW_FRACTION)
        ax.annotate("", xy=(end, y), xytext=(start, y), xycoords="axes fraction", textcoords="axes fraction",
                    arrowprops=dict(arrowstyle="-|>,head_width=0.13,head_length=0.32", color=MUTED_TEXT, lw=0.5,
                                    shrinkA=0, shrinkB=0))
        ax.text((start + end) / 2, y, f"{method} better", transform=ax.transAxes, fontsize=5.5, color=MUTED_TEXT,
                ha="center", va="center", bbox=dict(facecolor="white", edgecolor="none", pad=1), zorder=7)
for ax, letter in zip(axes, "ab"):
    box = ax.get_tightbbox().transformed(fig.transFigure.inverted())
    fig.text(0, box.y1, letter, fontsize=9, fontweight="bold", ha="left", va="top")

save_pub(fig, "flaviviridae_jackknife_difference", "figures")
plt.show()
```

## 7. Alternative figure: the differences with their jackknife confidence intervals

The same comparison as section 6, drawn as the test of section 5 sees it: for each protein, the difference of the trees
of all sequences and its 95% confidence interval, difference ± 1.96 jackknife standard errors. Unlike the replicate
boxes of section 6, whose spread is about half the uncertainty of the difference, an interval that covers the line at
no difference here means the difference is not significant (before the Holm correction, which only widens them).
Written to `figures/flaviviridae_jackknife_ci`.

```python
Z = stats.norm.ppf(0.975)
CI_COLOUR = BOX_COLOUR
CI_PANELS = [("Monophyly", "fraction", 100, "ESM3Di - AlphaFold2 (percentage points)", ("AlphaFold2", "ESM3Di")),
             ("Ultrametricity", "root_to_tip_cv", 1, "ESM3Di - AlphaFold2", ("ESM3Di", "AlphaFold2"))]

fig = plt.figure(figsize=(WIDTH_SINGLE, 70 * MM), layout="constrained")
axes = fig.subplots(len(CI_PANELS), 1)
for ax, (title, measure, factor, xlabel, _) in zip(axes, CI_PANELS):
    tests = significance[significance["measure"] == measure].set_index("protein").loc[list(DATASETS.values())]
    centre = factor * tests["difference"].to_numpy()
    half = factor * Z * tests["se"].to_numpy()
    rows = np.arange(len(tests))
    ax.plot([0, 0], [ARROW_ROW - 0.22, len(rows) - 0.55], color=MUTED_TEXT, lw=0.6, ls=(0, (2.5, 2)), zorder=1)
    ax.errorbar(centre, rows, xerr=half, fmt="none", ecolor=CI_COLOUR, elinewidth=1.2, capsize=2.5, capthick=0.8, zorder=3)
    for row, value in zip(rows, centre):
        ax.plot([value, value], [row - FULL_HALF, row + FULL_HALF], color="black", lw=0.9, solid_capstyle="butt", zorder=5)
    # symmetric about zero, so that both arrows fit and neither side looks favoured by the frame
    reach = 1.08 * np.max(np.abs(np.concatenate([centre - half, centre + half])))
    ax.set_xlim(-reach, reach)
    ax.set_ylim(len(rows) - 0.55, ARROW_ROW - 0.3)
    ax.set_yticks(rows, DATASETS.values(), fontsize=6)
    ax.tick_params(axis="y", length=0, pad=2)
    ax.spines["left"].set_visible(False)
    ax.set_xlabel(xlabel)
    ax.set_title(title, loc="left", pad=3)

axes[0].legend(handles=[Line2D([], [], color="black", lw=0.9, label="Tree of all sequences"),
                        Line2D([], [], color=CI_COLOUR, lw=1.2, label="95% CI (jackknife)"),
                        Line2D([], [], color=MUTED_TEXT, lw=0.6, ls=(0, (2.5, 2)), label="No difference")],
               loc="lower right", bbox_to_anchor=(1, 1.0), ncol=3, frameon=False, fontsize=5.5,
               handlelength=1.4, handletextpad=0.4, columnspacing=1.0, borderpad=0, borderaxespad=0.2)

fig.canvas.draw()
fig.set_layout_engine("none")
for ax, (*_, favoured) in zip(axes, CI_PANELS):
    zero = ax.transAxes.inverted().transform(ax.transData.transform((0, 0)))[0]
    y = ax.transAxes.inverted().transform(ax.transData.transform((0, ARROW_ROW)))[1]
    for side, method in zip((-1, 1), favoured):
        start, end = zero + side * 0.012, zero + side * (0.012 + ARROW_FRACTION)
        ax.annotate("", xy=(end, y), xytext=(start, y), xycoords="axes fraction", textcoords="axes fraction",
                    arrowprops=dict(arrowstyle="-|>,head_width=0.13,head_length=0.32", color=MUTED_TEXT, lw=0.5,
                                    shrinkA=0, shrinkB=0))
        ax.text((start + end) / 2, y, f"{method} better", transform=ax.transAxes, fontsize=5.5, color=MUTED_TEXT,
                ha="center", va="center", bbox=dict(facecolor="white", edgecolor="none", pad=1), zorder=7)
for ax, letter in zip(axes, "ab"):
    box = ax.get_tightbbox().transformed(fig.transFigure.inverted())
    fig.text(0, box.y1, letter, fontsize=9, fontweight="bold", ha="left", va="top")

save_pub(fig, "flaviviridae_jackknife_ci", "figures")
plt.show()
```
