```python
# count number of sequence sin fasta and extract length statistics
from Bio import SeqIO

# Count the number of sequences in the FASTA file
num_sequences = 0
for record in SeqIO.parse(fasta_path, "fasta"):
    num_sequences += 1

print(f"Number of sequences in : {num_sequences}")

# Extract length statistics
lengths = []
for record in SeqIO.parse(fasta_path, "fasta"):
    lengths.append(len(record.seq))

print(f"Length statistics for:")
print(f"  Min: {min(lengths)}")
print(f"  Max: {max(lengths)}")
print(f"  Mean: {sum(lengths) / len(lengths):.2f}")
```
