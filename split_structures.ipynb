```python
pdb_file_dir = "../Mifsud_et_al_Repository/glycoprotein_structural_alignments_and_trees/refolded_fullglyco"

pdb_file_list = [f for f in os.listdir(pdb_file_dir) if f.endswith(".pdb")]
```

```python
e_fasta_file = "../Mifsud_et_al_Repository/glycoprotein_structural_alignments_and_trees/3di/refolded_fullglyco_E_aa.fas"
e1_fasta_file = "../Mifsud_et_al_Repository/glycoprotein_structural_alignments_and_trees/3di/refolded_fullglyco_E1_aa.fas"
e2_fasta_file = "../Mifsud_et_al_Repository/glycoprotein_structural_alignments_and_trees/3di/refolded_fullglyco_E2_aa.fas"


fasta_files = [e_fasta_file, e1_fasta_file, e2_fasta_file]

# extract only header which is the pdb file name
fasta_headers = {}
for fasta_file in fasta_files:
    pdb_names = []
    with open(fasta_file, "r") as f:
        lines = f.readlines()
        for i in range(0, len(lines), 2):
            header = lines[i].strip()
            pdb_name = header.split(">")[1]  # Extract pdb name from header
            pdb_names.append(pdb_name)
    fasta_headers[fasta_file] = pdb_names
```

```python
set(pdb_file_list) == {pdb for v in fasta_headers.values() for pdb in v}
```

```python
# now for each fasta file create a folder containing the pdb files that are present in the fasta file
for fasta_file, pdb_names in fasta_headers.items():
    folder_name = os.path.splitext(os.path.basename(fasta_file))[0]  # Get the base name of the fasta file without extension
    print(folder_name)
    os.makedirs(folder_name, exist_ok=True)  # Create the folder if it doesn't exist

    for pdb_name in pdb_names:
        pdb_file_path = os.path.join(pdb_file_dir, pdb_name)  # Construct the full path to the pdb file
        if os.path.exists(pdb_file_path):  # Check if the pdb file exists
            # Copy the pdb file to the new folder
            os.system(f"cp {pdb_file_path} {folder_name}/")
```

```python
# number of files in each folder should be equal to the number of pdb names in the fasta file
print(f"Number of files in {folder_name}: {len(os.listdir(folder_name))}, Number of pdb names in fasta file: {len(pdb_names)}")
```
